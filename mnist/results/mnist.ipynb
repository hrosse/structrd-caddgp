```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm

import sys
from pathlib import Path

current = Path.cwd()
for parent in [current] + list(current.parents):
    if (parent / "utils").is_dir():
        sys.path.insert(0, str(parent))
        break

from utils.plot_setting import set_default_plots
```

```python
SAVE_FIG = True
```

```python
path = current / "mnist6000"

model_list = ["isotropic", "ARD", "APRD2x2", "APRD4x4", "APRD7x7"]
colour_list = [
    "#0072B2", "#E69F00", "#009E73",
    "#D55E00", "#CC79A7", "#000000"
]


score_dictlist = [dict() for m in model_list]
final_pred_list = [dict() for m in model_list]
model_param_list = [dict() for m in model_list]

for i in range(len(model_list)):
    with np.load(path / model_list[i] / "scores.npz") as data:
        score_dictlist[i]["model"] = model_list[i]
        score_dictlist[i]["step"] = data["step"]
        score_dictlist[i]["elapsed"] = data["elapsed"]
        score_dictlist[i]["elbo"] = -data["nelbo"]
        score_dictlist[i]["error"] = data["error"]
        score_dictlist[i]["nlpp"] = data["nlpp"]
        score_dictlist[i]["mean nlpp"] = data["nlpp"]/6000
        score_dictlist[i]["nlpp_miss"] = data["nlpp_miss"]
    with np.load(path / model_list[i] / "final_predictions.npz") as data:
        final_pred_list[i]["model"] = model_list[i]
        final_pred_list[i]["final_pred"] = data["final_pred"]
        final_pred_list[i]["final_pred_correct"] = data["final_pred_correct"]
        final_pred_list[i]["final_pred_miss"] = data["final_pred_miss"]
        final_pred_list[i]["correct_flag"] = data["correct_flag"]
    with np.load(path / model_list[i] / "model.npz") as data:
        model_param_list[i]["model"] = model_list[i]
        for key, value in data.items():
            model_param_list[i][key] =data[key]

score_dictlist6000 = score_dictlist
```

```python
# Plot Scores with respect to training steps
set_default_plots(figsize=(6.06, 4), dpi=300)
fig = plt.figure()
ax1 = fig.add_subplot(2,3,1)
for i in range(len(model_list)):
    ax1.plot(score_dictlist6000[i]["step"], score_dictlist6000[i]["elbo"], label=score_dictlist6000[i]["model"], color=colour_list[i])
ax1.set_xlabel("Step (in thousands)")
ax1.set_ylabel("ELBO")
ax1.set_xticks(range(0, 180001, 45000))
ax1.xaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, _: f'{x / 1000:.0f}')
)
ax2 = fig.add_subplot(2,3,2)
for i in range(len(model_list)):
    ax2.plot(score_dictlist6000[i]["step"], score_dictlist6000[i]["error"]*100, label=score_dictlist6000[i]["model"], color=colour_list[i])
ax2.set_xlabel("Step (in thousands)")
ax2.set_ylabel("Error (\%)")
ax2.set_xticks(range(0, 180001, 45000))
ax2.xaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, _: f'{x / 1000:.0f}')
)

ax3 = fig.add_subplot(2,3,3)
for i in range(len(model_list)):
    ax3.plot(score_dictlist6000[i]["step"], score_dictlist6000[i]["mean nlpp"]*100, label=score_dictlist6000[i]["model"], color=colour_list[i])
ax3.set_xlabel("Step (in thousands)")
ax3.set_ylabel("Mean NLPP")
ax3.set_xticks(range(0, 180001, 45000))
ax3.xaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, _: f'{x / 1000:.0f}')
)

########

ax4 = fig.add_subplot(2,3,4)
for i in range(len(model_list)):
    ax4.plot(score_dictlist60000[i]["step"], score_dictlist60000[i]["elbo"], label=score_dictlist60000[i]["model"], color=colour_list[i])
ax4.set_xlabel("Step (in thousands)")
ax4.set_ylabel("ELBO")
ax4.set_xticks(range(0, 300001, 60000))
ax4.xaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, _: f'{x / 1000:.0f}')
)


ax5 = fig.add_subplot(2,3,5)
for i in range(len(model_list)):
    ax5.plot(score_dictlist60000[i]["step"], score_dictlist60000[i]["error"]*100, label=score_dictlist60000[i]["model"], color=colour_list[i])
ax5.set_xlabel("Step (in thousands)")
ax5.set_ylabel("Error (\%)")
ax5.set_ylim(top=2.6)
ax5.set_xticks(range(0, 300001, 60000))
ax5.xaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, _: f'{x / 1000:.0f}')
)

ax6 = fig.add_subplot(2,3,6)
for i in range(len(model_list)):
    ax6.plot(score_dictlist60000[i]["step"], score_dictlist60000[i]["mean nlpp"]*100, label=score_dictlist60000[i]["model"], color=colour_list[i])
ax6.set_xlabel("Step (in thousands)")
ax6.set_ylabel("Mean NLPP")
ax6.set_ylim(top=1.6)
ax6.set_xticks(range(0, 300001, 60000))
ax6.xaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, _: f'{x / 1000:.0f}')
)


handles, labels = ax1.get_legend_handles_labels()
plt.subplots_adjust(bottom=0.22)
fig.legend(
    handles, labels,
    loc="lower center",
    ncol=5,
    bbox_to_anchor=(0.5, 0.05),
    fontsize=8
)

# Reserve space at the bottom for the legend
fig.tight_layout(rect=[0, 0.15, 1, 1])


if SAVE_FIG:
    plt.savefig('mnist_curves.pdf')

plt.show()
plt.close()
```
